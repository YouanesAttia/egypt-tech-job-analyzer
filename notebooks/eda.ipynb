```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ast
from collections import Counter

df = pd.read_csv('../output/jobs_cleaned.csv')
```

```python
df['skills_list'] = df['skills'].apply(ast.literal_eval)
all_skills = [skill for sublist in df['skills_list'] for skill in sublist]
skill_counts = Counter(all_skills).most_common(10)
skills_df = pd.DataFrame(skill_counts, columns=['Skill', 'Count'])

plt.figure(figsize=(10, 6))
sns.barplot(data=skills_df, x='Count', y='Skill')
plt.title('Market Dominance: Top 10 Most Demanded Skills in Egypt', fontsize=14)
plt.show()
```

Engineering is the undisputed leader with nearly 790 mentions, indicating a heavy industrial and structural focus in the current economic landscape. However, the high rankings of Sales (704) and Communication (608) reveal that "soft skills" and revenue-generating roles are critical gatekeepers for employment.

```python
plt.figure(figsize=(10, 6))
df['governorate'].value_counts().head(5).plot(kind='pie', autopct='%1.1f%%')
plt.title('Geographical Concentration of Tech Jobs', fontsize=14)
plt.ylabel('')
plt.show()
```

The visualization confirms a massive geographic bias. Typically, Cairo and Giza account for over 85% of all vacancies.


```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='experience_level', order=df['experience_level'].value_counts().index)
plt.title('Barrier to Entry: Distribution of Experience Levels')
plt.xticks(rotation=45)
plt.show()
```

Real numbers show that "Experienced" (Mid-level) roles often outweigh "Entry Level" roles by a ratio of 2.5:1. This highlights a gap in the Egyptian market, where companies prefer hiring proven talent rather than investing in fresh graduates, making the first job the hardest to secure.

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Duration')
plt.title('Employment Stability: Full-Time vs Freelance/Part-Time', fontsize=14)
plt.show()
```

Over 90% of postings are for "Full-time" positions. Egyptian tech employers still prioritize long-term, full-time contracts over freelance or project-based work, likely due to the need for intellectual property security and team stability.
